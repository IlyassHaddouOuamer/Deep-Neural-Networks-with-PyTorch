# Initialization with same weights

```python
import torch
import torch.nn as nn
from torch import sigmoid
import matplotlib.pylab as plt
import numpy as np
torch.manual_seed(0)
```

```python
# the function for plotting the model
def plotstuff(X, Y, model, epoch , leg=True):
    plt.plot(X.numpy(), model(X).detach().numpy(), label=('epoch' + str(epoch)))
    plt.plot(X.numpy(), Y.numpy(), 'r')
    plt.xlabel('x')
    if leg == True:
        plt.legend()
    else:
        pass
```

## Neural Network Model and traninig function

```python
class n_net(nn.Module):

    # Constructor
    def __init__(self, D_in, H, D_out):
        super(n_net, self).__init__()
        # hidden layer
        self.linear1 = nn.Linear(D_in, H)
        self.linear2 = nn.Linear(H, D_out)
        # Define the first linear layer
        self.a1 = None
        self.l1 = None
        self.l2 = None

    # Prediction
    def forward(self, x):
        self.l1 = self.linear1(x)
        self.a1 = sigmoid(self.l1)
        self.l2 = self.linear2(self.a1)
        yhat = sigmoid(self.l2)
        return yhat
```

```python
# Define the training function
def train(Y,X, model, optimizer, criterion, epochs = 1000):
    cost = []
    total = 0
    for epoch in range(epochs):
        total = 0
        for y,x in zip(Y,X):
            yhat = model(x)
            loss = criterion(yhat, y)
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()

            total += loss.item()

        cost.append(total)

        if epoch % 300 == 0:
            plotstuff(X,Y,model,epoch,leg=True)
            model(X)
            plt.scatter(model.a1.detach().numpy()[:,0], model.a1.detach().numpy()[:,1], c=Y.numpy().reshape(-1))
            plt.title('activations')


    return cost
```

## Make some Data

```python
X = torch.arange(-20,20,1).view(-1,1).type(torch.FloatTensor)
Y = torch.zeros(X.shape[0])
Y[(X[:, 0] > -4) & (X[:, 0] < 4)] = 1.0
```

### Define Neural Network with same weight initialization , define criterion function, optimizer and trainthe model

```python
# The loss function

def criterion_cross(output, labels):
    out = -1 * torch.mean(labels * torch.log(output) + (1 - labels) * torch.log(1 - output))
    return out
```

```python
# train the model
model = n_net(1, 2, 1)
```

```python
model.state_dict()
```

```python
model.state_dict()['linear1.weight'][0] = 1.0
model.state_dict()['linear1.weight'][1] = 1.0

model.state_dict()['linear1.bias'][0] = 0.0
model.state_dict()['linear1.bias'][1] = 0.0

model.state_dict()['linear2.weight'][0] = 1.0
model.state_dict()['linear2.bias'][0] = 0.0

model.state_dict()
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
train_model = train(Y, X, model, optimizer, criterion_cross, epochs=1000)

plt.plot(train_model)
plt.xlabel('epoch')
plt.title('Cross entropy loss')
```
